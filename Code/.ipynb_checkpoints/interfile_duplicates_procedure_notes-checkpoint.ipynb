```python
study = "Cancer"
folder = ["R3_3787_Lee_Cancer_Extract_Files", "R3_3787_Lee_Data_Cancer_20240509", "R3_3787_Lee_Data_Cancer_20250912"]
```

```python
final_unique_dfs = [] # List to hold the de-duplicated data from each file

# 2. Process each file
for i, file_name in enumerate(file_names):
    file_extension = os.path.splitext(file_name)[1].lower()
    file_path = os.path.join(shared_path, study, folder[i], file_name)
    
    if file_extension == '.csv':
        try:
            current_df = pd.read_csv(file_path, encoding='latin-1')
        except:
            print('file not exist in:', folder[i])
    elif file_extension == '.txt':
        try:
            current_df = pd.read_csv(file_path, sep='|')
        except:
            print('file not exist in:', folder[i])
    else:
        continue

    # --- KEY LOGIC FOR INTER-FILE DEDUPLICATION ---
    
    # Concatenate the current file's data with all previously processed data
    # (Do NOT use ignore_index=True here, we need separate indices for now)
    
    if final_unique_dfs:
        # Create a temporary DataFrame of ALL data processed so far
        all_prior_data = pd.concat(final_unique_dfs)
        
        # Check the current_df against all prior data. 
        # We only look at the columns that contain the actual data (excluding the index).
        data_columns = current_df.columns
        
        # Identify rows in the current file that are NOT duplicates of PRIOR files
        # The 'indicator=True' is used to identify the source of the merge
        merged = pd.merge(
            current_df, 
            all_prior_data, 
            on=list(data_columns), # Merge on all data columns
            how='left', 
            indicator=True
        )
        
        # Rows in the current file that are *only* in the 'left' (current_df) are unique
        unique_to_current_file = merged[merged['_merge'] == 'left_only'].drop(columns='_merge')
        
        # Append the unique data (which includes any intra-file duplicates)
        final_unique_dfs.append(unique_to_current_file)
    else:
        # The first file is added as is (it has no prior files to check against)
        final_unique_dfs.append(current_df)


# 3. Final Concatenation
if final_unique_dfs:
    # Concatenate the final list of DataFrames to produce the result
    final_df = pd.concat(final_unique_dfs, ignore_index=True)

    print("✅ Successfully merged and removed ONLY inter-file duplicates.")
    print("\nFinal DataFrame head (includes intra-file duplicates):")
else:
    print("❌ No valid files were read. The final DataFrame is empty.")
```

## **Extract Ki-67 info**
Ki-67 was then categorized to grades G1 (≤2 %), G2 (3–20 %), or G3 (>20 %) according to the European Neuroendocrine Tumor Society (ENETS) guidelines and the 2010 World Health Organization (WHO) classification. 

### <span style="color:blue"> **Ki-67**</span> check existence

```python
ki_67 = ['Ki-67', 'Ki67']
# study_diag = ["DIAG", "DIAGNOSTIC", "DX", "Diagnostic"]
```

```python
# Assuming your DataFrame is named 'df' and the column where you found the words is 'Notes'
column_to_check = 'NOTE_TEXT' 
type_column = 'Ki-67'

# Ensure the column is treated as strings and handle missing values
unique_df_aggregated[column_to_check] = unique_df_aggregated[column_to_check].astype(str)

# ---Create the OR-Condition Mask ---
# The pattern uses the '|' (OR) symbol.
# We use .str.contains() with case=False to handle both "DIAG" and "Diagnostic".
terms = [term for term in ki_67]
pattern = '|'.join([re.escape(term) for term in terms])

# Create a boolean mask: True if the string contains either word (case-insensitive)
mask = unique_df_aggregated[column_to_check].str.contains(pattern, case=False, na=False, regex=True)

# --- Assign the value "DIAG" to the 'type' column ---
# Use .loc to assign the value only where the mask is True
unique_df_aggregated.loc[mask, type_column] = 1
```

### <span style="color:blue"> **Ki-67**</span> check %

```python
column_to_check = 'NOTE_TEXT'
percentage_column = 'Ki-67_Percent'

# 1. Ensure the column is treated as strings
unique_df_aggregated[column_to_check] = unique_df_aggregated[column_to_check].astype(str)

# 2. Construct the robust regex pattern
# Pattern: (Ki-67 term) (Any chars except % non-greedily) (CAPTURED number with optional decimal) (%)
terms_pattern = '|'.join([re.escape(term) for term in ki_67])
# regex = rf'(?:{terms_pattern})[^%]*?(\d+\.?\d*)\s*\%'
# regex = rf'(?:{terms_pattern}).*?(\d+\.?\d*)\s*(?:%|\.|\b|\))'
# regex = rf'(?:{terms_pattern})\s*(?:of|approximately\s*)?(\d+\.?\d*)\s*(?:%|\.|\b)'


# 2. Construct the regex:
# Pattern: (Ki-67 term) (Optional Approximator/Connector term) (CAPTURED number) (optional terminator)
# approximation_terms = r'(?:of|approximately|around|is|index at|=|index of|at|ranging  between|labeling index:|is is low  at|low of|index is borderline at|high at|up to|very high at|proliferation index of| greater than|  index of greater than|index |index: |over|at\s*least)\s*' 
# regex = rf'(?:Ki\s*-\s*67|Ki67)\s*(?:{approximation_terms})?(\d+\.?\d*)\s*(?:%|\.|\b)'

ki67_synonyms = r'(?:Ki\s*-\s*67|Ki67)'
bridge = r'[^0-9.,]*?'
number_capture = r'(\d+\.?\d*)'
optional_percent = r'\s*%?'
regex = rf'{ki67_synonyms}{bridge}{number_capture}{optional_percent}'

# approximators = r'(?:of|approximately|around|is|index|at|ranging|between|labeling|low|high|up to|greater than|>|<|>=|<=|:|=)'
# regex = rf'{ki67_synonyms}.*?{approximators}?\s*(\d+\.?\d*)\s*%[^0-9]*'

# 3. Extract the percentage number using the regex
unique_df_aggregated[percentage_column] = unique_df_aggregated[column_to_check].str.extract(regex, expand=False, flags=re.IGNORECASE)


# 4. Convert the extracted column to a numeric type, coercing errors to NaN
# This ensures the values are ready for calculations and handles non-matches (NaN).
unique_df_aggregated[percentage_column] = pd.to_numeric(unique_df_aggregated[percentage_column], errors='coerce')
```

### <span style="color:blue"> **Ki-67**</span> check grade level (low, medium, high), and assign grade (G1, G2, G3)
Ki-67: low, medium, high

```python
levels = ['low', 'moderate', 'high']

# Combine the words into a single regex pattern for capture
level_term = '|'.join([re.escape(term) for term in levels])

ki67_synonyms = r'(?:Ki\s*-\s*67|Ki67)'
bridge = r'[^.,]*?'
regex_word_after = rf'{ki67_synonyms}{bridge}\b({level_term})\b'
regex_word_before = rf'\b({level_term})\b{bridge}{ki67_synonyms}'

# # Regex to find word AFTER Ki-67 (e.g., "Ki-67 is moderate")
# # regex_word_after = rf'(?:Ki-67|Ki67)(?: of| is| index is| proliferation index is| very|with| is very|of )?\s*({level_term})\b'
# regex_word_after = rf'(?:Ki\s*-\s*67|Ki67)\s*(?: of| is| index is| proliferation index is|proliferation index is very| very|index|with| is very|of )?\s*({level_term})\b'


# # Regex to find word BEFORE Ki-67 (e.g., "low Ki-67")
# # Note: The 'low' in the second note is captured here.
# regex_word_before = rf'({level_term})\s*[,]?\s*(?:Ki-67|Ki67)\b'

# Extract the word, prioritizing the word after Ki-67
unique_df_aggregated['Ki-67_WordLevel'] = unique_df_aggregated['NOTE_TEXT'].str.extract(regex_word_after, flags=re.IGNORECASE).fillna(
    unique_df_aggregated['NOTE_TEXT'].str.extract(regex_word_before, flags=re.IGNORECASE)
)

unique_df_aggregated['Ki-67_WordLevel'] = unique_df_aggregated['Ki-67_WordLevel'].str.lower()
```

### <span style="color:blue"> **Ki-67**</span> assign final level
Ki-67 was then categorized to grades low (≤2 %), moderate (3–20 %), or high (>20 %)

```python
unique_df_aggregated['Ki-67_Grade'] = None
```
