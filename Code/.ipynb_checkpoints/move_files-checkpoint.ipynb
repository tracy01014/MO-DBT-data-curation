```python
import pandas as pd
import os
import numpy as np
import shutil
from pathlib import Path
from tqdm import tqdm
```

```python
pd.set_option('display.max_columns', None)
```

#### <span style="color:#FF6347;">**READ**</span> mo_cancer_cohort_focus

```python
file_path = os.path.join("../Data/", "Cancer", 'mo_cancer_cohort_focus' + ".xlsx")
df = pd.read_excel(file_path)
```

#### <span style="color:blue;">**Main**</span>: construct target path and move files from source path (folder_path) to target path

```python
df
```

```python
df_copy = df.copy()
```

```python
root_path = 'P:/Dataset/MO-DBT'
SEP = os.sep

sec_1 = df_copy['subtype']
sec_2 = df_copy['PATIENT_STUDY_ID'].astype(str, errors='ignore')
sec_3 = df_copy['side'] + df_copy['view'] + '_' + df_copy['series']

df_copy['target_path'] = root_path + SEP + sec_1 + SEP + sec_2 + SEP + sec_3 
```

```python
df['target_path'] = df_copy['target_path']
```

```python
df['folder_path'][3]
```

```python
df['target_path'][3]
```

```python
df_mo_cancer = df[~df['target_path'].isna()]
```

```python
df_mo_cancer
```

#### <span style="color:#FF6347;">**SAVE**</span> **mo_cancer_cohort_focus**

```python
output_file = os.path.join("../../Subtype-Classification/Data/", 'MO-DBT-label-path' + ".xlsx")
df_mo_cancer.to_excel(output_file, index=False)
```

#### <span style="color:blue;">**Main**</span>: continue

```python
for _, row in tqdm(df_mo_cancer.iterrows(), total=len(df_mo_cancer), desc='move files'):
    source_dir = Path(row['folder_path'])
    target_dir = Path(row['target_path'])
            
    if not source_dir.is_dir():
        print(f"WARNING: Source directory not found: {source_dir}")
        continue

    if not target_dir.is_dir():
        try:
            target_dir.mkdir(parents=True, exist_ok=True)
        except Exception as e:
            print(f"  CRITICAL ERROR: Could not create target directory {target_dir.name}: {e}")
            continue
        
    # Iterate over ALL files within the source directory
    for item in source_dir.iterdir():
        source_item = item # The full path of the file/folder to move
        target_item = target_dir / item.name # The full path where it will land

        # print(source_item, target_item)

        try:
            # This handles both files and directories
            shutil.move(source_item, target_item)
            # print(f"  MOVED: {item.name}")
            
        except Exception as e:
            print(f"  ERROR moving {item.name}: {e}")
```
